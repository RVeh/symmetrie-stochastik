# Python-Programme

Die drei Notebooks ergänzen den Beitrag
*Was hat ein Platz, was ein anderer nicht hat?* um Simulationen und
empirische Untersuchungen.

Die veränderbaren Eingaben stehen jeweils zu Beginn. Mit `seed = 42`
sind die Simulationen reproduzierbar.

## Auswahl

1. [Gleichberechtigung der Positionen](Simulation_Urne_ohne_Zuruecklegen.ipynb)  
   Relative Häufigkeiten $h(A_j)$, ihre Stabilisierung und ein Ausblick
   auf die hypergeometrische Verteilung.

2. [Positionen weißer Kugeln](Simulation_erwartete_Positionen.ipynb)  
   Empirische Erwartungswerte, Varianzen und Standardabweichungen der
   Zufallsgrößen $T_j$ sowie die Spiegelsymmetrie.

3. [Lückenlängen](Simulation_Luecken.ipynb)  
   Empirische und exakte Verteilung einer Lücke, Kenngrößen und
   punktweise Wilson-Konfidenzintervalle.

## Leitgedanke

> Die Simulation lässt Strukturen sichtbar werden, die Kombinatorik
> liefert exakte Ergebnisse -- die Symmetrie erklärt sie.

Die Programme ersetzen die theoretischen Argumentationen nicht. Sie
machen Variabilität sichtbar, eröffnen Fragen und ermöglichen die
Kontrolle exakter Ergebnisse.
